```python

import pandas as pd
from pathlib import Path

project_dir = Path.cwd()

if project_dir.name == "notebooks":
    project_dir = project_dir.parent

data_dir = project_dir / "data"

train = pd.read_csv(data_dir / "train.csv")
test = pd.read_csv(data_dir / "test.csv")
sample = pd.read_csv(data_dir / "sample_submission.csv")

print("Datasets loaded successfully!")
```

```python

print("Training dataset:", train.shape)
print("Testing dataset:", test.shape)
print("Submission dataset:", sample.shape)
```

```python

print("Column names:")
print(train.columns.tolist())

print("\nFirst 5 transactions:")
display(train.head())
```

```python

print(train["is_fraud"].value_counts())
```

```python

total = len(train)

legitimate = (train["is_fraud"] == 0).sum()
fraud = (train["is_fraud"] == 1).sum()

legitimate_pct = legitimate / total * 100
fraud_pct = fraud / total * 100

print(f"Total transactions: {total}")
print(f"Legitimate: {legitimate_pct:.2f}%")
print(f"Fraud: {fraud_pct:.2f}%")
```
